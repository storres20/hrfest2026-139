```python
# ================================================================
# HRFEST 2026 - PAPER 03
# MULTI-NODE HOSPITAL COLD-CHAIN MONITORING
#
# Complete reproducible data-continuity analysis
#
# Includes:
#  1. Robust DataGrip/MongoDB CSV reader
#  2. Raw dataset characterization
#  3. Installation/configuration phase characterization
#  4. Common nominal 1-min persistence window
#  5. Duplicate-slot detection
#  6. Persistence availability
#  7. Temperature-data availability
#  8. Missing-slot detection
#  9. Gap-event detection
# 10. Inter-record interval analysis
# 11. Simultaneous three-node availability
# 12. Daily availability
# 13. Publication-quality figures
# 14. CSV result export
# 15. Reviewer-ready summary
# 16. ZIP download
# ================================================================


# ================================================================
# 1. LIBRARIES
# ================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from google.colab import files
from IPython.display import display
import zipfile
import csv
import os


pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 300)
pd.set_option("display.width", 250)


# ================================================================
# 2. EXPERIMENT CONFIGURATION
# ================================================================

NODES = [
    "ULT26-001",
    "MHT2025-001",
    "BLD26-001"
]

APPLICATIONS = {
    "ULT26-001": "Plasma freezer",
    "MHT2025-001": "Blood-storage refrigerator",
    "BLD26-001": "Reagent/sample refrigerator"
}


# ----------------------------------------------------------------
# START OF NOMINAL ONE-MINUTE PERSISTENCE REGIME
#
# Records before this timestamp correspond to physical installation
# and configuration activities.
#
# They are retained in the raw dataset for transparency but are
# excluded from the primary continuity analysis because the final
# one-minute persistence regime had not yet been established.
# ----------------------------------------------------------------

ANALYSIS_START = pd.Timestamp(
    "2026-08-21 19:07:00",
    tz="UTC"
)


# ================================================================
# 3. OUTPUT DIRECTORIES
# ================================================================

OUTPUT_DIR = Path("/content/results")
FIG_DIR = OUTPUT_DIR / "figures"

OUTPUT_DIR.mkdir(
    exist_ok=True
)

FIG_DIR.mkdir(
    exist_ok=True
)


print("=" * 80)
print("MULTI-NODE HOSPITAL COLD-CHAIN CONTINUITY ANALYSIS")
print("=" * 80)

print("\nConfigured nodes:")

for node in NODES:

    print(
        f"  {node}: "
        f"{APPLICATIONS[node]}"
    )

print(
    "\nNominal one-minute persistence analysis start:"
)

print(
    ANALYSIS_START
)


# ================================================================
# 4. UPLOAD RAW CSV
# ================================================================

print("\nUpload the RAW CSV exported from MongoDB/DataGrip.")

uploaded = files.upload()

filename = next(
    iter(uploaded)
)

print(
    "\nUploaded file:",
    filename
)


# ================================================================
# 5. ROBUST DATAGRIP / MONGODB CSV READER
#
# The exported CSV contains two row structures:
# - rows with the 9 fields declared in the header
# - rows with 12 fields
#
# The additional three fields are preserved with neutral names.
# They are NOT used in the continuity analysis because their
# semantic meaning is not defined by the CSV header.
# ================================================================

with open(
    filename,
    "r",
    encoding="utf-8-sig",
    newline=""
) as f:

    reader = csv.reader(f)

    original_header = next(reader)

    original_header = [
        h.strip()
        for h in original_header
    ]

    HEADER_FIELDS = len(original_header)

    raw_rows = []

    field_count_distribution = {}

    max_fields = HEADER_FIELDS

    for line_number, row in enumerate(
        reader,
        start=2
    ):

        n_fields = len(row)

        field_count_distribution[n_fields] = (
            field_count_distribution.get(n_fields, 0) + 1
        )

        max_fields = max(
            max_fields,
            n_fields
        )

        raw_rows.append(
            (line_number, row)
        )


print("\n" + "=" * 80)
print("CSV STRUCTURE INSPECTION")
print("=" * 80)

print(
    "Header fields:",
    HEADER_FIELDS
)

print(
    "Original header:",
    original_header
)

print(
    "Maximum fields detected in a row:",
    max_fields
)

print(
    "\nField-count distribution:"
)

for n_fields, count in sorted(
    field_count_distribution.items()
):

    print(
        f"  {n_fields} fields: "
        f"{count:,} rows"
    )


# ================================================================
# 6. PRESERVE ADDITIONAL UNLABELED FIELDS
# ================================================================

extra_count = (
    max_fields - HEADER_FIELDS
)


extended_header = (
    original_header.copy()
)


if extra_count > 0:

    for i in range(
        1,
        extra_count + 1
    ):

        extended_header.append(
            f"extra_field_{i}"
        )


print(
    "\nExtended internal header:"
)

print(
    extended_header
)


normalized_rows = []

short_rows = []


for line_number, row in raw_rows:

    # If a row contains fewer fields than the maximum structure,
    # preserve it and fill only the absent trailing positions.

    if len(row) < max_fields:

        short_rows.append({

            "line_number":
                line_number,

            "fields_found":
                len(row),

            "fields_added":
                max_fields - len(row)

        })

        row = row + [""] * (
            max_fields - len(row)
        )

    normalized_rows.append(
        row
    )


df_raw = pd.DataFrame(
    normalized_rows,
    columns=extended_header
)


df_raw.columns = (
    df_raw.columns
    .str.strip()
)


print("\n" + "=" * 80)
print("CSV LOAD VALIDATION")
print("=" * 80)

print(
    "Rows successfully loaded:",
    f"{len(df_raw):,}"
)

print(
    "Internal columns:",
    len(df_raw.columns)
)

print(
    "\nColumns:"
)

print(
    df_raw.columns.tolist()
)


# ================================================================
# 6A. CHARACTERIZE EXTRA FIELDS
# ================================================================

extra_columns = [

    col
    for col in df_raw.columns

    if col.startswith(
        "extra_field_"
    )
]


if extra_columns:

    print(
        "\nAdditional unlabeled fields detected:"
    )

    print(
        extra_columns
    )

    print(
        "\nNon-empty values in additional fields:"
    )

    for col in extra_columns:

        nonempty = (

            df_raw[col]
            .astype(str)
            .str.strip()
            .ne("")
            .sum()
        )

        print(
            f"  {col}: "
            f"{nonempty:,}"
        )


    print(
        "\nExample rows containing additional data:"
    )

    extra_nonempty_mask = np.zeros(
        len(df_raw),
        dtype=bool
    )

    for col in extra_columns:

        extra_nonempty_mask |= (

            df_raw[col]
            .astype(str)
            .str.strip()
            .ne("")
            .values
        )


    display(

        df_raw.loc[
            extra_nonempty_mask,
            original_header + extra_columns
        ]

        .head(10)
    )


# Save complete normalized dataset for audit purposes.

df_raw.to_csv(

    OUTPUT_DIR /
    "normalized_raw_dataset_with_extra_fields.csv",

    index=False
)


# Save a description of the CSV structure.

structure_rows = []

for n_fields, count in sorted(
    field_count_distribution.items()
):

    structure_rows.append({

        "number_of_fields":
            n_fields,

        "number_of_rows":
            count
    })


pd.DataFrame(
    structure_rows
).to_csv(

    OUTPUT_DIR /
    "csv_structure_summary.csv",

    index=False
)


print(
    "\nIMPORTANT:"
)

print(
    "The additional unlabeled fields are preserved "
    "but will NOT be used in the continuity analysis."
)

print(
    "Continuity analysis uses only explicitly named "
    "fields from the original CSV header."
)

# ================================================================
# 7. REQUIRED COLUMN VALIDATION
# ================================================================

required_columns = [
    "username",
    "createdAt",
    "time_slot"
]

missing_required = [

    col
    for col in required_columns

    if col not in df_raw.columns
]


if missing_required:

    raise ValueError(

        "Required columns missing: "
        f"{missing_required}"
    )


# ================================================================
# 8. TIMESTAMP CONVERSION
# ================================================================

timestamp_columns = [
    "createdAt",
    "updatedAt",
    "datetime",
    "time_slot"
]


for col in timestamp_columns:

    if col in df_raw.columns:

        df_raw[col] = pd.to_datetime(
            df_raw[col],
            utc=True,
            errors="coerce"
        )


print("\n" + "=" * 80)
print("TIMESTAMP VALIDATION")
print("=" * 80)

print(
    "Invalid createdAt:",
    df_raw["createdAt"]
    .isna()
    .sum()
)

print(
    "Invalid time_slot:",
    df_raw["time_slot"]
    .isna()
    .sum()
)


# ================================================================
# 9. FILTER STUDY NODES
# ================================================================

study_raw = df_raw[
    df_raw["username"]
    .isin(NODES)
].copy()


print("\n" + "=" * 80)
print("RAW STUDY DATASET")
print("=" * 80)

print(
    "Study-node records:",
    f"{len(study_raw):,}"
)


raw_node_counts = (
    study_raw["username"]
    .value_counts()
    .reindex(NODES)
)


print(
    "\nRecords per node:"
)

print(
    raw_node_counts
)


# ================================================================
# 10. RAW TEMPORAL RANGE
# ================================================================

raw_ranges = (

    study_raw

    .groupby(
        "username"
    )

    .agg(

        first_createdAt=(
            "createdAt",
            "min"
        ),

        last_createdAt=(
            "createdAt",
            "max"
        ),

        first_time_slot=(
            "time_slot",
            "min"
        ),

        last_time_slot=(
            "time_slot",
            "max"
        ),

        raw_records=(
            "createdAt",
            "size"
        )
    )

    .reindex(
        NODES
    )
)


print(
    "\nRAW TEMPORAL RANGE"
)

display(
    raw_ranges
)


raw_ranges.to_csv(
    OUTPUT_DIR /
    "raw_temporal_ranges.csv"
)


# ================================================================
# 11. INITIAL INSTALLATION / CONFIGURATION PHASE
# ================================================================

initial_phase = study_raw[
    study_raw["createdAt"]
    < ANALYSIS_START
].copy()


initial_summary = (

    initial_phase

    .groupby(
        "username"
    )

    .agg(

        records=(
            "createdAt",
            "size"
        ),

        first_record=(
            "createdAt",
            "min"
        ),

        last_record=(
            "createdAt",
            "max"
        ),

        first_slot=(
            "time_slot",
            "min"
        ),

        last_slot=(
            "time_slot",
            "max"
        )
    )

    .reindex(
        NODES
    )
)


print("\n" + "=" * 80)
print("INITIAL INSTALLATION / CONFIGURATION PHASE")
print("=" * 80)

print(
    "\nRecords from this phase are retained "
    "in the raw dataset but excluded from "
    "the primary one-minute continuity analysis."
)

display(
    initial_summary
)


initial_phase.to_csv(
    OUTPUT_DIR /
    "initial_configuration_phase.csv",
    index=False
)


initial_summary.to_csv(
    OUTPUT_DIR /
    "initial_configuration_summary.csv"
)


# ================================================================
# 12. INITIAL-PHASE INTER-RECORD INTERVALS
# ================================================================

initial_interval_frames = []


for node in NODES:

    tmp = (

        initial_phase[
            initial_phase["username"]
            == node
        ]

        .sort_values(
            "createdAt"
        )

        .copy()
    )


    if len(tmp) > 0:

        tmp[
            "inter_record_min"
        ] = (

            tmp["createdAt"]
            .diff()
            .dt
            .total_seconds()
            / 60
        )


        initial_interval_frames.append(

            tmp[
                [
                    "username",
                    "createdAt",
                    "time_slot",
                    "inter_record_min"
                ]
            ]
        )


if initial_interval_frames:

    initial_intervals = pd.concat(
        initial_interval_frames,
        ignore_index=True
    )

else:

    initial_intervals = pd.DataFrame()


initial_intervals.to_csv(
    OUTPUT_DIR /
    "initial_phase_intervals.csv",
    index=False
)


# ================================================================
# 13. POST-CONFIGURATION CANDIDATE DATA
# ================================================================

candidate = study_raw[
    study_raw["createdAt"]
    >= ANALYSIS_START
].copy()


first_records = (

    candidate

    .groupby(
        "username"
    )["createdAt"]

    .min()

    .reindex(
        NODES
    )
)


last_records = (

    candidate

    .groupby(
        "username"
    )["createdAt"]

    .max()

    .reindex(
        NODES
    )
)


print("\n" + "=" * 80)
print("POST-CONFIGURATION RANGE")
print("=" * 80)

print(
    "\nFirst record after configuration:"
)

display(
    first_records
)

print(
    "\nLast available record:"
)

display(
    last_records
)


if last_records.isna().any():

    raise ValueError(

        "\nAt least one study node has "
        "no post-configuration records."
    )


# ================================================================
# 14. COMMON THREE-NODE EXPERIMENTAL WINDOW
#
# The common observation period ends when the first of the three
# nodes stops producing records.
# ================================================================

COMMON_END = (
    last_records.min()
)


START_SLOT = (
    ANALYSIS_START.floor(
        "min"
    )
)


END_SLOT = (
    COMMON_END.floor(
        "min"
    )
)


expected_slots = pd.date_range(

    start=START_SLOT,

    end=END_SLOT,

    freq="1min",

    tz="UTC"
)


N_EXPECTED = len(
    expected_slots
)


# Number of one-minute slots includes
# both start and end slots.

duration_elapsed = (
    END_SLOT - START_SLOT
)


duration_hours = (
    duration_elapsed.total_seconds()
    / 3600
)


duration_days = (
    duration_elapsed.total_seconds()
    / 86400
)


print("\n" + "=" * 80)
print("COMMON THREE-NODE EXPERIMENTAL WINDOW")
print("=" * 80)

print(
    "Start:",
    START_SLOT
)

print(
    "End:",
    END_SLOT
)

print(
    "Expected one-minute slots per node:",
    f"{N_EXPECTED:,}"
)

print(
    "Elapsed duration:",
    duration_elapsed
)

print(
    "Elapsed hours:",
    f"{duration_hours:.3f}"
)

print(
    "Elapsed days:",
    f"{duration_days:.4f}"
)


# ================================================================
# 15. CREATE EXPERIMENTAL DATASET
# ================================================================

df = candidate[
    candidate["createdAt"]
    <= COMMON_END
].copy()


# Nominal persistence slot

df["slot"] = (

    df["time_slot"]
    .dt.floor(
        "min"
    )
)


# Keep only nominal slots inside
# common analysis window

df = df[

    (df["slot"] >= START_SLOT)

    &

    (df["slot"] <= END_SLOT)

].copy()


df = df.sort_values(

    [
        "username",
        "createdAt"
    ]

).reset_index(
    drop=True
)


print("\nExperimental records:")

print(
    f"{len(df):,}"
)


print(
    "\nExperimental records per node:"
)

print(

    df["username"]

    .value_counts()

    .reindex(
        NODES
    )
)


# ================================================================
# 16. DUPLICATE NODE / SLOT CHECK
# ================================================================

duplicate_mask = df.duplicated(

    subset=[
        "username",
        "slot"
    ],

    keep=False
)


duplicates = df[
    duplicate_mask
].copy()


duplicate_groups = (

    df

    .groupby(
        [
            "username",
            "slot"
        ]
    )

    .size()

    .reset_index(
        name="records_in_slot"
    )
)


duplicate_groups = duplicate_groups[

    duplicate_groups[
        "records_in_slot"
    ] > 1

].copy()


print("\n" + "=" * 80)
print("DUPLICATE SLOT CHECK")
print("=" * 80)

print(
    "Rows involved in duplicated slots:",
    len(duplicates)
)

print(
    "Duplicated node/slot groups:",
    len(duplicate_groups)
)


if len(
    duplicate_groups
) > 0:

    display(
        duplicate_groups.head(
            50
        )
    )


duplicates.to_csv(
    OUTPUT_DIR /
    "duplicate_slot_rows.csv",
    index=False
)


duplicate_groups.to_csv(
    OUTPUT_DIR /
    "duplicate_slot_groups.csv",
    index=False
)


# ================================================================
# 17. ONE RECORD PER NOMINAL SLOT
#
# For availability, a minute counts only once.
# If duplicates exist, the last persisted record is retained
# for temperature completeness analysis.
# ================================================================

slot_df = (

    df

    .sort_values(
        "createdAt"
    )

    .drop_duplicates(

        subset=[
            "username",
            "slot"
        ],

        keep="last"
    )

    .copy()
)


# ================================================================
# 18. TEMPERATURE COLUMN IDENTIFICATION
# ================================================================

temperature_candidates = [

    "dsTemperature",
    "temperature",
    "Temperature",
    "temp",
    "temperatureValue"
]


TEMP_COL = None


for col in temperature_candidates:

    if col in slot_df.columns:

        TEMP_COL = col

        break


print("\nTemperature column:")

print(
    TEMP_COL
)


if TEMP_COL is None:

    print(
        "\nWARNING: Temperature column "
        "could not be identified."
    )


# ================================================================
# 19. CONVERT TEMPERATURE TO NUMERIC
#
# Empty strings and non-numeric values become NaN.
# ================================================================

if TEMP_COL is not None:

    slot_df[
        TEMP_COL
    ] = pd.to_numeric(

        slot_df[
            TEMP_COL
        ],

        errors="coerce"
    )


# ================================================================
# 20. INDIVIDUAL PERSISTENCE + TEMPERATURE AVAILABILITY
# ================================================================

summary_rows = []


for node in NODES:

    node_df = slot_df[
        slot_df["username"]
        == node
    ].copy()


    stored_slots = (

        node_df["slot"]
        .nunique()
    )


    missing_slots = (

        N_EXPECTED
        -
        stored_slots
    )


    persistence_availability = (

        100.0
        *
        stored_slots
        /
        N_EXPECTED
    )


    if TEMP_COL is not None:

        null_temperature = (

            node_df[
                TEMP_COL
            ]

            .isna()

            .sum()
        )


        valid_temperature = (

            node_df[
                TEMP_COL
            ]

            .notna()

            .sum()
        )


        temperature_availability = (

            100.0
            *
            valid_temperature
            /
            N_EXPECTED
        )

    else:

        null_temperature = np.nan

        valid_temperature = np.nan

        temperature_availability = np.nan


    summary_rows.append({

        "node":
            node,

        "application":
            APPLICATIONS[node],

        "expected_slots":
            N_EXPECTED,

        "stored_slots":
            stored_slots,

        "missing_slots":
            missing_slots,

        "persistence_availability_pct":
            persistence_availability,

        "null_temperature":
            null_temperature,

        "valid_temperature":
            valid_temperature,

        "temperature_availability_pct":
            temperature_availability
    })


summary = pd.DataFrame(
    summary_rows
)


# ================================================================
# 21. GAP EVENT DETECTION
#
# Definition:
#
# A gap event consists of one or more consecutive nominal
# one-minute slots for which no persistent record exists.
#
# This definition is independent from createdAt jitter.
# ================================================================

gap_frames = []


for node in NODES:

    node_slots = set(

        slot_df.loc[

            slot_df["username"]
            == node,

            "slot"

        ].dropna()
    )


    missing = [

        slot

        for slot in expected_slots

        if slot not in node_slots
    ]


    if len(
        missing
    ) == 0:

        continue


    missing_df = pd.DataFrame({

        "slot":
            pd.DatetimeIndex(
                missing
            )
    })


    missing_df[
        "difference"
    ] = (

        missing_df[
            "slot"
        ]

        .diff()
    )


    missing_df[
        "new_gap"
    ] = (

        missing_df[
            "difference"
        ]

        .ne(
            pd.Timedelta(
                minutes=1
            )
        )
    )


    missing_df[
        "gap_id"
    ] = (

        missing_df[
            "new_gap"
        ]

        .cumsum()
    )


    gaps = (

        missing_df

        .groupby(
            "gap_id"
        )

        .agg(

            gap_start=(
                "slot",
                "min"
            ),

            gap_end=(
                "slot",
                "max"
            ),

            missing_slots=(
                "slot",
                "size"
            )
        )

        .reset_index(
            drop=True
        )
    )


    gaps.insert(
        0,
        "node",
        node
    )


    gap_frames.append(
        gaps
    )


if gap_frames:

    gaps_all = pd.concat(

        gap_frames,

        ignore_index=True
    )

else:

    gaps_all = pd.DataFrame(

        columns=[

            "node",
            "gap_start",
            "gap_end",
            "missing_slots"
        ]
    )


print("\n" + "=" * 80)
print("MISSING-SLOT GAP EVENTS")
print("=" * 80)

if len(
    gaps_all
) > 0:

    display(
        gaps_all
    )

else:

    print(
        "No missing-slot gap events detected."
    )


gaps_all.to_csv(

    OUTPUT_DIR /
    "gap_events.csv",

    index=False
)


# ================================================================
# 22. GAP SUMMARY PER NODE
# ================================================================

gap_summary_rows = []


for node in NODES:

    node_gaps = gaps_all[
        gaps_all["node"]
        == node
    ]


    gap_events = len(
        node_gaps
    )


    if gap_events > 0:

        max_missing_run = int(

            node_gaps[
                "missing_slots"
            ]

            .max()
        )

    else:

        max_missing_run = 0


    gap_summary_rows.append({

        "node":
            node,

        "gap_events":
            gap_events,

        "max_missing_slot_run_min":
            max_missing_run
    })


gap_summary = pd.DataFrame(
    gap_summary_rows
)


summary = summary.merge(

    gap_summary,

    on="node",

    how="left"
)


# ================================================================
# 23. INTER-RECORD INTERVAL ANALYSIS USING createdAt
#
# IMPORTANT:
#
# createdAt interval and missing nominal slots are different
# metrics.
#
# createdAt captures actual persistence timing/jitter.
# time_slot captures nominal one-minute continuity.
# ================================================================

interval_rows = []


for node in NODES:

    node_df = (

        df[
            df["username"]
            == node
        ]

        .sort_values(
            "createdAt"
        )

        .copy()
    )


    intervals = (

        node_df[
            "createdAt"
        ]

        .diff()

        .dt
        .total_seconds()

        .div(
            60
        )

        .dropna()
    )


    interval_rows.append({

        "node":
            node,

        "median_inter_record_min":
            intervals.median(),

        "mean_inter_record_min":
            intervals.mean(),

        "std_inter_record_min":
            intervals.std(),

        "max_inter_record_min":
            intervals.max()
    })


interval_summary = pd.DataFrame(
    interval_rows
)


summary = summary.merge(

    interval_summary,

    on="node",

    how="left"
)


# ================================================================
# 24. FINAL INDIVIDUAL-NODE TABLE
# ================================================================

final_columns = [

    "node",
    "application",

    "expected_slots",
    "stored_slots",
    "missing_slots",

    "persistence_availability_pct",

    "gap_events",
    "max_missing_slot_run_min",

    "median_inter_record_min",
    "mean_inter_record_min",
    "std_inter_record_min",
    "max_inter_record_min",

    "null_temperature",
    "valid_temperature",
    "temperature_availability_pct"
]


final_results = summary[
    final_columns
].copy()


print("\n" + "=" * 80)
print("INDIVIDUAL NODE RESULTS")
print("=" * 80)

display(

    final_results.round(
        4
    )
)


# ================================================================
# 25. CREATE AVAILABILITY MATRIX
# ================================================================

availability_matrix = pd.DataFrame(

    index=expected_slots
)


availability_matrix.index.name = (
    "time_slot"
)


slot_sets = {}


for node in NODES:

    slot_sets[
        node
    ] = set(

        slot_df.loc[

            slot_df["username"]
            == node,

            "slot"

        ].dropna()
    )


    availability_matrix[
        node
    ] = [

        1
        if slot in slot_sets[node]
        else 0

        for slot in expected_slots
    ]


# ================================================================
# 26. SIMULTANEOUS THREE-NODE PERSISTENCE AVAILABILITY
# ================================================================

simultaneous_slots = (

    slot_sets[
        NODES[0]
    ]

    &

    slot_sets[
        NODES[1]
    ]

    &

    slot_sets[
        NODES[2]
    ]
)


N_SIMULTANEOUS = len(
    simultaneous_slots
)


SIMULTANEOUS_AVAILABILITY = (

    100.0
    *
    N_SIMULTANEOUS
    /
    N_EXPECTED
)


print("\n" + "=" * 80)
print("SIMULTANEOUS THREE-NODE PERSISTENCE")
print("=" * 80)

print(
    "Complete simultaneous slots:",
    f"{N_SIMULTANEOUS:,}"
)

print(
    "Expected slots:",
    f"{N_EXPECTED:,}"
)

print(
    "Simultaneous persistence availability:",
    f"{SIMULTANEOUS_AVAILABILITY:.4f}%"
)


# ================================================================
# 27. SIMULTANEOUS VALID-TEMPERATURE AVAILABILITY
# ================================================================

N_SIM_VALID_TEMP = np.nan

SIM_VALID_TEMP_AVAILABILITY = np.nan


if TEMP_COL is not None:

    valid_temperature_sets = {}


    for node in NODES:

        node_df = slot_df[

            slot_df["username"]
            == node

        ].copy()


        valid_temperature_sets[
            node
        ] = set(

            node_df.loc[

                node_df[
                    TEMP_COL
                ].notna(),

                "slot"

            ]
        )


    simultaneous_valid_temperature = (

        valid_temperature_sets[
            NODES[0]
        ]

        &

        valid_temperature_sets[
            NODES[1]
        ]

        &

        valid_temperature_sets[
            NODES[2]
        ]
    )


    N_SIM_VALID_TEMP = len(

        simultaneous_valid_temperature
    )


    SIM_VALID_TEMP_AVAILABILITY = (

        100.0
        *
        N_SIM_VALID_TEMP
        /
        N_EXPECTED
    )


    print("\n" + "=" * 80)
    print("SIMULTANEOUS THREE-NODE VALID TEMPERATURE")
    print("=" * 80)

    print(
        "Simultaneous valid-temperature slots:",
        f"{N_SIM_VALID_TEMP:,}"
    )

    print(
        "Expected slots:",
        f"{N_EXPECTED:,}"
    )

    print(
        "Simultaneous valid-temperature availability:",
        f"{SIM_VALID_TEMP_AVAILABILITY:.4f}%"
    )


# ================================================================
# 28. DAILY PERSISTENCE AVAILABILITY
#
# Useful for Reviewer 2:
# Shows continuity over the extended observation period instead
# of reporting only an aggregate percentage.
# ================================================================

daily_frames = []


for node in NODES:

    node_daily = pd.DataFrame({

        "time_slot":
            expected_slots,

        "stored":
            availability_matrix[
                node
            ].values
    })


    node_daily[
        "date"
    ] = (

        node_daily[
            "time_slot"
        ]

        .dt
        .date
    )


    daily = (

        node_daily

        .groupby(
            "date"
        )

        .agg(

            expected_slots=(
                "stored",
                "size"
            ),

            stored_slots=(
                "stored",
                "sum"
            )
        )

        .reset_index()
    )


    daily[
        "missing_slots"
    ] = (

        daily[
            "expected_slots"
        ]

        -

        daily[
            "stored_slots"
        ]
    )


    daily[
        "availability_pct"
    ] = (

        100.0

        *

        daily[
            "stored_slots"
        ]

        /

        daily[
            "expected_slots"
        ]
    )


    daily[
        "node"
    ] = node


    daily_frames.append(
        daily
    )


daily_availability = pd.concat(

    daily_frames,

    ignore_index=True
)


print("\n" + "=" * 80)
print("DAILY PERSISTENCE AVAILABILITY")
print("=" * 80)

display(
    daily_availability.round(
        4
    )
)


# ================================================================
# 29. DAILY SIMULTANEOUS THREE-NODE AVAILABILITY
# ================================================================

simultaneous_daily = pd.DataFrame({

    "time_slot":
        expected_slots
})


simultaneous_daily[
    "complete_three_nodes"
] = (

    availability_matrix[
        NODES
    ]

    .min(
        axis=1
    )

    .values
)


simultaneous_daily[
    "date"
] = (

    simultaneous_daily[
        "time_slot"
    ]

    .dt
    .date
)


daily_simultaneous = (

    simultaneous_daily

    .groupby(
        "date"
    )

    .agg(

        expected_slots=(
            "complete_three_nodes",
            "size"
        ),

        simultaneous_slots=(
            "complete_three_nodes",
            "sum"
        )
    )

    .reset_index()
)


daily_simultaneous[
    "missing_simultaneous_slots"
] = (

    daily_simultaneous[
        "expected_slots"
    ]

    -

    daily_simultaneous[
        "simultaneous_slots"
    ]
)


daily_simultaneous[
    "simultaneous_availability_pct"
] = (

    100.0

    *

    daily_simultaneous[
        "simultaneous_slots"
    ]

    /

    daily_simultaneous[
        "expected_slots"
    ]
)


print("\n" + "=" * 80)
print("DAILY SIMULTANEOUS THREE-NODE AVAILABILITY")
print("=" * 80)

display(

    daily_simultaneous.round(
        4
    )
)


# ================================================================
# 30. FIGURE 1
# FULL COMMON-WINDOW DATA CONTINUITY
# ================================================================

fig, ax = plt.subplots(

    figsize=(
        16,
        5
    )
)


offsets = {

    "ULT26-001": 2,
    "MHT2025-001": 1,
    "BLD26-001": 0
}


for node in NODES:

    present = (

        availability_matrix[
            node
        ]

        == 1
    )


    ax.scatter(

        availability_matrix.index[
            present
        ],

        np.full(

            present.sum(),

            offsets[
                node
            ]
        ),

        s=2,

        label=node
    )


ax.set_yticks(
    [0, 1, 2]
)


ax.set_yticklabels([

    "BLD26-001",
    "MHT2025-001",
    "ULT26-001"
])


ax.set_xlabel(
    "Date and time (UTC)"
)


ax.set_ylabel(
    "Monitoring node"
)


ax.set_title(

    "Persistent Data Continuity During the Common "
    "One-Minute Observation Period"
)


ax.grid(

    axis="x",

    alpha=0.3
)


plt.tight_layout()


FIG1 = (

    FIG_DIR /
    "fig1_full_continuity.png"
)


plt.savefig(

    FIG1,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


# ================================================================
# 31. FIGURE 2
# MISSING ONE-MINUTE PERSISTENCE SLOTS
# ================================================================

fig, ax = plt.subplots(

    figsize=(
        16,
        4
    )
)


for node in NODES:

    missing_mask = (

        availability_matrix[
            node
        ]

        == 0
    )


    missing_times = (

        availability_matrix.index[
            missing_mask
        ]
    )


    if len(
        missing_times
    ) > 0:

        ax.scatter(

            missing_times,

            np.full(

                len(
                    missing_times
                ),

                offsets[
                    node
                ]
            ),

            s=14,

            label=node
        )


ax.set_yticks(
    [0, 1, 2]
)


ax.set_yticklabels([

    "BLD26-001",
    "MHT2025-001",
    "ULT26-001"
])


ax.set_xlabel(
    "Date and time (UTC)"
)


ax.set_ylabel(
    "Monitoring node"
)


ax.set_title(

    "Missing One-Minute Persistence Slots "
    "During Concurrent Operation"
)


ax.grid(

    axis="x",

    alpha=0.3
)


plt.tight_layout()


FIG2 = (

    FIG_DIR /
    "fig2_missing_slots.png"
)


plt.savefig(

    FIG2,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


# ================================================================
# 32. FIGURE 3
# INDIVIDUAL PERSISTENCE AVAILABILITY
# ================================================================

fig, ax = plt.subplots(

    figsize=(
        8,
        5
    )
)


bars = ax.bar(

    final_results[
        "node"
    ],

    final_results[
        "persistence_availability_pct"
    ]
)


ax.set_ylabel(
    "Persistence availability (%)"
)


ax.set_xlabel(
    "Monitoring node"
)


ax.set_title(
    "Individual-Node Persistence Availability"
)


minimum_availability = (

    final_results[
        "persistence_availability_pct"
    ]

    .min()
)


ax.set_ylim(

    max(
        0,
        minimum_availability - 0.5
    ),

    100.05
)


for bar, value in zip(

    bars,

    final_results[
        "persistence_availability_pct"
    ]
):

    ax.text(

        bar.get_x()
        +
        bar.get_width() / 2,

        value + 0.01,

        f"{value:.2f}%",

        ha="center",

        va="bottom"
    )


plt.tight_layout()


FIG3 = (

    FIG_DIR /
    "fig3_persistence_availability.png"
)


plt.savefig(

    FIG3,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


# ================================================================
# 33. FIGURE 4
# TEMPERATURE-DATA AVAILABILITY
# ================================================================

if TEMP_COL is not None:

    fig, ax = plt.subplots(

        figsize=(
            8,
            5
        )
    )


    bars = ax.bar(

        final_results[
            "node"
        ],

        final_results[
            "temperature_availability_pct"
        ]
    )


    ax.set_ylabel(
        "Valid-temperature availability (%)"
    )


    ax.set_xlabel(
        "Monitoring node"
    )


    ax.set_title(
        "Temperature-Data Availability"
    )


    minimum_temperature_availability = (

        final_results[
            "temperature_availability_pct"
        ]

        .min()
    )


    ax.set_ylim(

        max(

            0,

            minimum_temperature_availability
            - 1

        ),

        100.05
    )


    for bar, value in zip(

        bars,

        final_results[
            "temperature_availability_pct"
        ]
    ):

        ax.text(

            bar.get_x()
            +
            bar.get_width() / 2,

            value + 0.02,

            f"{value:.2f}%",

            ha="center",

            va="bottom"
        )


    plt.tight_layout()


    FIG4 = (

        FIG_DIR /
        "fig4_temperature_availability.png"
    )


    plt.savefig(

        FIG4,

        dpi=300,

        bbox_inches="tight"
    )


    plt.show()


# ================================================================
# 34. FIGURE 5
# DAILY PERSISTENCE AVAILABILITY
#
# Important for Reviewer 2 because it shows behavior throughout
# the longer common observation period.
# ================================================================

fig, ax = plt.subplots(

    figsize=(
        14,
        6
    )
)


for node in NODES:

    tmp = daily_availability[

        daily_availability[
            "node"
        ]

        == node
    ]


    ax.plot(

        pd.to_datetime(
            tmp["date"]
        ),

        tmp[
            "availability_pct"
        ],

        marker="o",

        label=node
    )


ax.set_xlabel(
    "Date (UTC)"
)


ax.set_ylabel(
    "Daily persistence availability (%)"
)


ax.set_title(

    "Daily Persistence Availability During "
    "the Common Multi-Node Observation Period"
)


ax.legend()


ax.grid(
    alpha=0.3
)


plt.tight_layout()


FIG5 = (

    FIG_DIR /
    "fig5_daily_availability.png"
)


plt.savefig(

    FIG5,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


# ================================================================
# 35. FIGURE 6
# DAILY SIMULTANEOUS THREE-NODE AVAILABILITY
# ================================================================

fig, ax = plt.subplots(

    figsize=(
        14,
        5
    )
)


ax.plot(

    pd.to_datetime(
        daily_simultaneous[
            "date"
        ]
    ),

    daily_simultaneous[
        "simultaneous_availability_pct"
    ],

    marker="o"
)


ax.set_xlabel(
    "Date (UTC)"
)


ax.set_ylabel(
    "Simultaneous availability (%)"
)


ax.set_title(

    "Daily Simultaneous Three-Node "
    "Persistence Availability"
)


ax.grid(
    alpha=0.3
)


plt.tight_layout()


FIG6 = (

    FIG_DIR /
    "fig6_daily_simultaneous_availability.png"
)


plt.savefig(

    FIG6,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


# ================================================================
# 36. EXPORT RESULT TABLES
# ================================================================

final_results.to_csv(

    OUTPUT_DIR /
    "node_summary.csv",

    index=False
)


availability_matrix.to_csv(

    OUTPUT_DIR /
    "one_minute_availability_matrix.csv"
)


daily_availability.to_csv(

    OUTPUT_DIR /
    "daily_availability.csv",

    index=False
)


daily_simultaneous.to_csv(

    OUTPUT_DIR /
    "daily_simultaneous_availability.csv",

    index=False
)


# ================================================================
# 37. EXPERIMENTAL WINDOW SUMMARY
# ================================================================

window_summary = pd.DataFrame([{

    "analysis_start_utc":
        START_SLOT,

    "analysis_end_utc":
        END_SLOT,

    "expected_one_minute_slots_per_node":
        N_EXPECTED,

    "elapsed_hours":
        duration_hours,

    "elapsed_days":
        duration_days,

    "simultaneous_persistence_slots":
        N_SIMULTANEOUS,

    "simultaneous_persistence_availability_pct":
        SIMULTANEOUS_AVAILABILITY,

    "simultaneous_valid_temperature_slots":
        N_SIM_VALID_TEMP,

    "simultaneous_valid_temperature_availability_pct":
        SIM_VALID_TEMP_AVAILABILITY
}])


window_summary.to_csv(

    OUTPUT_DIR /
    "experimental_window_summary.csv",

    index=False
)


# ================================================================
# 38. SAVE CLEAN EXPERIMENTAL DATASET
#
# IMPORTANT:
# This does NOT replace the raw dataset.
# It is a derived dataset containing the defined experimental
# analysis window.
# ================================================================

slot_df.to_csv(

    OUTPUT_DIR /
    "derived_experimental_dataset.csv",

    index=False
)


# ================================================================
# 39. REVIEWER-READY TEXT SUMMARY
# ================================================================

summary_lines = []


def add_line(text=""):

    print(text)

    summary_lines.append(
        str(text)
    )


print("\n\n")

add_line(
    "=" * 90
)

add_line(
    "REVIEWER-READY ANALYSIS SUMMARY"
)

add_line(
    "=" * 90
)


add_line(
    "\n[A] RAW DATASET"
)

add_line(
    f"Study-node raw records: "
    f"{len(study_raw):,}"
)


add_line(
    "\n[B] INITIAL INSTALLATION / CONFIGURATION PHASE"
)

add_line(
    initial_summary.to_string()
)


add_line(
    "\n[C] COMMON NOMINAL ONE-MINUTE EXPERIMENTAL WINDOW"
)

add_line(
    f"Start: {START_SLOT}"
)

add_line(
    f"End: {END_SLOT}"
)

add_line(
    f"Expected one-minute slots per node: "
    f"{N_EXPECTED:,}"
)

add_line(
    f"Elapsed duration: "
    f"{duration_elapsed}"
)

add_line(
    f"Elapsed hours: "
    f"{duration_hours:.3f}"
)

add_line(
    f"Elapsed days: "
    f"{duration_days:.4f}"
)


add_line(
    "\n[D] INDIVIDUAL NODE RESULTS"
)

add_line(

    final_results

    .round(
        4
    )

    .to_string(
        index=False
    )
)


add_line(
    "\n[E] GAP EVENTS"
)


if len(
    gaps_all
) > 0:

    add_line(

        gaps_all

        .to_string(
            index=False
        )
    )

else:

    add_line(
        "No missing-slot gap events detected."
    )


add_line(
    "\n[F] SIMULTANEOUS THREE-NODE PERSISTENCE"
)

add_line(

    f"{N_SIMULTANEOUS:,}"
    f"/"
    f"{N_EXPECTED:,}"
    f" = "
    f"{SIMULTANEOUS_AVAILABILITY:.4f}%"
)


if TEMP_COL is not None:

    add_line(
        "\n[G] SIMULTANEOUS THREE-NODE VALID TEMPERATURE"
    )

    add_line(

        f"{N_SIM_VALID_TEMP:,}"
        f"/"
        f"{N_EXPECTED:,}"
        f" = "
        f"{SIM_VALID_TEMP_AVAILABILITY:.4f}%"
    )


add_line(
    "\n[H] DUPLICATE SLOT CHECK"
)

add_line(

    f"Duplicated node/slot groups: "
    f"{len(duplicate_groups):,}"
)


add_line(
    "\n[I] CSV PARSER VALIDATION"
)

add_line(

    f"Rows with additional EMPTY trailing fields: "
    f"{len(extra_field_rows):,}"
)

add_line(

    f"Rows with fewer fields: "
    f"{len(short_rows):,}"
)


add_line(
    "\n[J] REVIEWER 2 - SCOPE INFORMATION"
)

add_line(

    f"Common three-node observation duration: "
    f"{duration_days:.4f} days"
)

add_line(

    f"Number of physical monitoring nodes: "
    f"{len(NODES)}"
)

add_line(

    "NOTE: The extended temporal observation improves "
    "the duration of the experimental evaluation, but "
    "the deployment still contains only three physical "
    "nodes and therefore should not be interpreted as "
    "evidence of large-scale scalability."
)


add_line(
    "\n[K] REVIEWER 2 - FAULT ATTRIBUTION"
)

add_line(

    "NOTE: Missing persistent records represent "
    "end-to-end data-continuity interruptions. "
    "The persistent database dataset alone cannot "
    "attribute a gap to the sensing node, Wi-Fi, "
    "WebSocket connection, backend processing, or "
    "database writing."
)


add_line(
    "\n[L] GENERATED FILES"
)


for path in sorted(

    OUTPUT_DIR.rglob(
        "*"
    )
):

    if path.is_file():

        add_line(
            str(path)
        )


add_line(
    "=" * 90
)


# ================================================================
# 40. SAVE REVIEWER SUMMARY AS TXT
# ================================================================

summary_txt = (

    OUTPUT_DIR /
    "reviewer_ready_summary.txt"
)


with open(

    summary_txt,

    "w",

    encoding="utf-8"

) as f:

    f.write(

        "\n".join(
            summary_lines
        )
    )


# ================================================================
# 41. CREATE README-LIKE ANALYSIS DESCRIPTION
# ================================================================

methodology_text = f"""
MULTI-NODE COLD-CHAIN DATA-CONTINUITY ANALYSIS
==============================================

Primary analysis start:
{START_SLOT}

Common analysis end:
{END_SLOT}

Expected nominal one-minute slots per node:
{N_EXPECTED}

Analysis rationale
------------------
Records generated before {START_SLOT} correspond to the initial
installation and configuration phase. During this phase, the final
nominal one-minute database persistence regime had not yet been
established.

These initial records were retained in the raw dataset for
transparency but excluded from the primary continuity analysis.

Persistence availability
------------------------
Persistence availability is calculated as:

stored nominal one-minute slots / expected nominal one-minute slots

A minute is counted as available when at least one persistent record
exists for that node and nominal time slot.

Temperature-data availability
-----------------------------
Temperature-data availability is calculated as:

nominal slots containing a numeric temperature value /
expected nominal one-minute slots

Therefore, a persisted record with an empty or non-numeric temperature
value contributes to persistence availability but not to
temperature-data availability.

Gap event
---------
A gap event is defined as one or more consecutive nominal one-minute
slots without a persistent record.

Inter-record intervals
----------------------
Inter-record intervals are calculated independently using createdAt.
These intervals describe actual persistence timing and should not be
confused with the number of missing nominal time slots.

Simultaneous availability
-------------------------
Simultaneous three-node persistence availability represents the
fraction of expected nominal time slots in which all three monitoring
nodes contain a persistent record.

Fault attribution limitation
----------------------------
The persistent database records permit end-to-end continuity
assessment but do not permit causal attribution of a missing record
to a specific subsystem such as the sensing node, Wi-Fi connection,
WebSocket communication, backend processing, or MongoDB persistence.

Scale limitation
----------------
The extended observation period strengthens the temporal evaluation,
but the experiment still contains three physical monitoring nodes.
The results therefore characterize this deployment and should not be
interpreted as evidence of large-scale scalability.
"""


with open(

    OUTPUT_DIR /
    "analysis_methodology.txt",

    "w",

    encoding="utf-8"

) as f:

    f.write(
        methodology_text
    )


# ================================================================
# 42. CREATE ZIP WITH ALL RESULTS
# ================================================================

ZIP_PATH = (

    "/content/"
    "multinode_analysis_results.zip"
)


with zipfile.ZipFile(

    ZIP_PATH,

    "w",

    zipfile.ZIP_DEFLATED

) as zipf:


    for file_path in (

        OUTPUT_DIR.rglob(
            "*"
        )
    ):

        if file_path.is_file():

            zipf.write(

                file_path,

                arcname=file_path.relative_to(
                    OUTPUT_DIR
                )
            )


print("\n" + "=" * 80)
print("ANALYSIS COMPLETED SUCCESSFULLY")
print("=" * 80)

print(
    "\nResults ZIP:"
)

print(
    ZIP_PATH
)

print(
    "\nDownload will start automatically."
)


# ================================================================
# 43. DOWNLOAD ZIP
# ================================================================

files.download(
    ZIP_PATH
)
```
